```python
import pandas as pd
```

```python
pdb_info  = pd.read_csv(
                "INDEX_general_PL_name.2020",
                delimiter=r"\s{2,}",
                engine="python",
                skiprows=6,
                header=None,
                on_bad_lines="skip",
            )
pdb_info.columns = [
    "PDB",
    "Year",
    "Uniprot ID",
    "PDB_Name"
]
```

```python
pdb_info.head()
```

```python
# 读取三个 CSV 文件
train_df = pd.read_csv('../raw2016/train.csv')
val_df = pd.read_csv('../raw2016/valid.csv')
test_df = pd.read_csv('../raw2016/test.csv')

# 为每个 DataFrame 添加 source 列
train_df['Source'] = 'train'
val_df['Source'] = 'val'
test_df['Source'] = 'test'

# 合并三个 DataFrame
combined_df = pd.concat([train_df, val_df, test_df], ignore_index=True)
combined_df = combined_df.rename(columns={'regression_label': 'Label'})
combined_df = combined_df.drop(columns=['Target_Chain'])
```

```python
combined_df
```

```python
# 假设你的主 DataFrame 名为 df
combined_df['PDB'] = combined_df['ID'].str[-4:].str.lower()
pdb_info['PDB'] = pdb_info['PDB'].str.lower()
```

```python
combined_df.head()
```

```python
pdb_info.head()
```

```python
combined_df.shape
```

```python
combined_df = combined_df.merge(pdb_info[['PDB', 'Uniprot ID']], on='PDB', how='left')
```

```python
combined_df = combined_df.drop(columns=['PDB'])
```

```python
combined_df.head()
```

```python
combined_df.drop(columns=['Target_ID'], inplace=True)
```

```python
# 保存合并后的 DataFrame 到一个新的 CSV 文件
combined_df.to_csv('../full.csv', index=False)
```
